```python
reviews_df = pd.read_csv('mgames/games_reviews.csv')
games_df   = pd.read_csv('mgames/games.csv')

print("=== ДО очистки ===")
print(f"reviews: {reviews_df.shape}, games: {games_df.shape}")
print(f"Колонки reviews: {list(reviews_df.columns)}")
print(f"Колонки games:   {list(games_df.columns)}")


reviews_df['id'] = reviews_df['id'].astype(int)
games_df['id']   = games_df['id'].astype(int)


print("\nreview_type до фильтрации:")
print(reviews_df['review_type'].value_counts(dropna=False))

reviews_df = reviews_df[reviews_df['review_type'] == 'user'].copy()

print("\nreview_type после фильтрации:")
print(reviews_df['review_type'].value_counts(dropna=False))

reviews_df = reviews_df.drop(
    columns=['review_type', 'platform', 'date', 'publicationName'],
    errors='ignore'
)

reviews_df['score'] = pd.to_numeric(reviews_df['score'], errors='coerce')

key_cols = ['author', 'id', 'quote', 'score']
reviews_df = reviews_df.dropna(subset=key_cols)

reviews_df['quote'] = reviews_df['quote'].astype(str).str.strip()
reviews_df = reviews_df[reviews_df['quote'].str.len() >= 5]

print("\n=== Распределение rating ===")
print(reviews_df['score'].describe())
print("\nУникальные значения score:")
print(sorted(reviews_df['score'].unique()))

before = len(reviews_df)
reviews_df = reviews_df.drop_duplicates(subset=['author', 'id'], keep='last')
print(f"\nУдалено дубликатов (author, id): {before - len(reviews_df)}")

print("\n=== ПОСЛЕ очистки ===")
n_users = reviews_df['author'].nunique()
n_games = reviews_df['id'].nunique()
n_reviews = len(reviews_df)

print(f"Юзеров:  {n_users}")
print(f"Игр:     {n_games}")
print(f"Отзывов: {n_reviews}")
print(f"Плотность: {n_reviews / (n_users * n_games):.6f}")
```

```python

RANDOM_STATE = 42


def proportional_split(df, test_size=0.2, min_test=1, seed=42):
    rng = np.random.RandomState(seed)
    df = df.copy()
    df['_rand'] = rng.rand(len(df))
    df = df.sort_values(['author', '_rand'])

    counts = df.groupby('author')['id'].transform('size')
    df['_rank'] = df.groupby('author').cumcount(ascending=False)

    target_test = np.maximum(min_test, (counts * test_size).round().astype(int))
    target_test = np.minimum(target_test, counts - 1)

    test_mask = (df['_rank'] < target_test) & (counts > min_test)

    test_df  = df[test_mask].drop(columns=['_rand', '_rank']).reset_index(drop=True)
    train_df = df[~test_mask].drop(columns=['_rand', '_rank']).reset_index(drop=True)

    return train_df, test_df

train_df, test_df = proportional_split(reviews_df, test_size=0.2,
                                        min_test=1, seed=RANDOM_STATE)


print("=== SPLIT ===")
print(f"train: {len(train_df):>8d} отзывов | "
      f"{train_df['author'].nunique():>6d} юзеров | "
      f"{train_df['id'].nunique():>6d} игр")
print(f"test:  {len(test_df):>8d} отзывов | "
      f"{test_df['author'].nunique():>6d} юзеров | "
      f"{test_df['id'].nunique():>6d} игр")


common_users = set(train_df['author']) & set(test_df['author'])
print(f"Юзеров в обоих наборах: {len(common_users)}")


test_only_users = set(test_df['author']) - set(train_df['author'])
print(f"Юзеров только в test:   {len(test_only_users)}")

test_counts = test_df.groupby('author').size()
print("\n=== Распределение test-отзывов на юзера ===")
print(test_counts.value_counts().sort_index())
```

```python

class HybridCFRecommender:
    def __init__(self, k_user=50, k_item=50, alpha=0.5,
                 score_min=0.0, score_max=10.0):
        self.k_user = k_user
        self.k_item = k_item
        self.alpha = alpha
        self.score_min = score_min
        self.score_max = score_max

    def fit(self, train_df):
        self.users_ = np.array(sorted(train_df['author'].unique()))
        self.items_ = np.array(sorted(train_df['id'].unique()))
        self.uidx_ = {u: i for i, u in enumerate(self.users_)}
        self.iidx_ = {it: i for i, it in enumerate(self.items_)}
        self.n_users_ = len(self.users_)
        self.n_items_ = len(self.items_)

        rows = train_df['author'].map(self.uidx_).to_numpy()
        cols = train_df['id'].map(self.iidx_).to_numpy()
        vals = train_df['score'].astype(np.float32).to_numpy()

        self.R = csr_matrix((vals, (rows, cols)),
                            shape=(self.n_users_, self.n_items_))
        self.M = self.R.copy()
        self.M.data = np.ones_like(self.M.data)

        self.global_mean_ = float(vals.mean())
        self.user_means_ = (
            np.asarray(self.R.sum(1)).ravel()
            / np.maximum(np.asarray(self.M.sum(1)).ravel(), 1)
        )
        self.user_train_counts_ = self.M.sum(1).A1.astype(int)

        self.Rc = csr_matrix(
            (vals - self.user_means_[rows], (rows, cols)),
            shape=self.R.shape,
        )

        self.S_user = self._topk_sim(self.Rc, self.k_user) 
        self.S_item = self._topk_sim(self.Rc.T.tocsr(), self.k_item) 
        self.S_item_T = self.S_item.T.tocsr()
        self.abs_S_item_T = self._abs_sparse(self.S_item_T)

        return self

    @staticmethod
    def _abs_sparse(S):
        S2 = S.copy()
        S2.data = np.abs(S2.data)
        return S2

    @staticmethod
    def _topk_sim(X, k):
        n = X.shape[0]
        k = min(k, n - 1)
        if k <= 0:
            return csr_matrix((n, n))
        nn = NearestNeighbors(n_neighbors=k + 1, metric='cosine',
                              algorithm='brute', n_jobs=-1)
        nn.fit(X)
        dist, idx = nn.kneighbors(X)
        rows = np.repeat(np.arange(n), k + 1)
        cols = idx.ravel()
        sims = 1.0 - dist.ravel()
        m = rows != cols  # убираем самого себя
        return csr_matrix((sims[m], (rows[m], cols[m])), shape=(n, n))

    def score_row(self, u_idx):
        mean_u = self.user_means_[u_idx]

        nbrs = self.S_user[u_idx].indices
        sims = self.S_user[u_idx].data
        if len(nbrs) > 0:
            sub_Rc = self.Rc[nbrs]
            sub_M  = self.M[nbrs]
            num_u = np.asarray(sub_Rc.T.dot(sims)).ravel()
            den_u = np.asarray(sub_M.T.dot(np.abs(sims))).ravel()
            pred_u = mean_u + np.divide(
                num_u, den_u,
                out=np.zeros_like(num_u, dtype=np.float32),
                where=den_u > 0,
            )
            pred_u[den_u == 0] = mean_u
        else:
            pred_u = np.full(self.n_items_, mean_u, dtype=np.float32)

        row_Rc = self.Rc[u_idx]
        row_M  = self.M[u_idx]
        if row_Rc.nnz > 0:
            num_i = np.asarray((row_Rc @ self.S_item_T).todense()).ravel()
            den_i = np.asarray((row_M  @ self.abs_S_item_T).todense()).ravel()
            pred_i = mean_u + np.divide(
                num_i, den_i,
                out=np.zeros_like(num_i, dtype=np.float32),
                where=den_i > 0,
            )
            pred_i[den_i == 0] = mean_u
        else:
            pred_i = np.full(self.n_items_, mean_u, dtype=np.float32)

        if self.alpha >= 1.0:
            pred = pred_u
        elif self.alpha <= 0.0:
            pred = pred_i
        else:
            pred = self.alpha * pred_u + (1.0 - self.alpha) * pred_i

        np.clip(pred, self.score_min, self.score_max, out=pred)
        return pred

    def recommend(self, user_ids, top_k=10, exclude_seen=True):
        result = {}
        for u in user_ids:
            ui = self.uidx_.get(u, -1)
            if ui < 0:
                continue
            scores = self.score_row(ui)
            if exclude_seen:
                scores[self.M[ui].indices] = -np.inf
            n = min(top_k, self.n_items_)
            top = np.argpartition(-scores, n - 1)[:n]
            top = top[np.argsort(-scores[top])]
            result[u] = [(self.items_[i], float(scores[i])) for i in top]
        return result

def evaluate_cf(recommender, test_df, k=10, threshold=3.0,
                stratum_by='train'):
    test_rel = test_df[test_df['score'] >= threshold]
    rel_by_user = test_rel.groupby('author')['id'].apply(set).to_dict()
    if not rel_by_user:
        return pd.DataFrame()

    def _stratum(c):
        if c <= 1:  return 'cold (1)'
        if c <= 4:  return 'warm (2-4)'
        return 'active (5+)'

    valid_users = [u for u in rel_by_user if u in recommender.uidx_]
    recs = recommender.recommend(valid_users, top_k=k)

    if stratum_by == 'train':
        cnt_map = {u: int(recommender.user_train_counts_[recommender.uidx_[u]])
                   for u in valid_users}
    else: 
        cnt_map = test_df.groupby('author').size().to_dict()


    log2 = np.log2
    discounts = 1.0 / log2(np.arange(2, k + 2))

    rows = []
    for u in valid_users:
        rel = rel_by_user[u]
        rec_items = [it for it, _ in recs.get(u, [])]
        hits = np.array([it in rel for it in rec_items])

        n_hits = int(hits.sum())
        precision = n_hits / k
        hit_rate = 1.0 if n_hits > 0 else 0.0

        if n_hits > 0:
            first_rank = int(np.argmax(hits)) + 1
            mrr = 1.0 / first_rank
        else:
            mrr = 0.0

        dcg = float(np.sum(hits * discounts))
        idcg = float(np.sum(discounts[:min(len(rel), k)]))
        ndcg = dcg / idcg if idcg > 0 else 0.0

        rows.append({
            'author': u,
            'stratum': _stratum(cnt_map.get(u, 0)),
            'precision': precision,
            'hit_rate': hit_rate,
            'mrr': mrr,
            'ndcg': ndcg,
        })

    df_res = pd.DataFrame(rows)

    agg = df_res.groupby('stratum').agg(**{
        f'precision@{k}': ('precision', 'mean'),
        f'hit_rate@{k}':  ('hit_rate',  'mean'),
        f'mrr@{k}':       ('mrr',       'mean'),
        f'ndcg@{k}':      ('ndcg',      'mean'),
        'n_users':        ('author',    'count'),
    })

    all_row = pd.DataFrame({
        f'precision@{k}': [df_res['precision'].mean()],
        f'hit_rate@{k}':  [df_res['hit_rate'].mean()],
        f'mrr@{k}':       [df_res['mrr'].mean()],
        f'ndcg@{k}':      [df_res['ndcg'].mean()],
        'n_users':        [len(df_res)],
    }, index=['all'])

    result = pd.concat([agg, all_row])
    order = ['cold (1)', 'warm (2-4)', 'active (5+)', 'all']
    result = result.reindex([x for x in order if x in result.index])
    return result


model = HybridCFRecommender(
    k_user=50, k_item=50, alpha=0.3,
    score_min=float(train_df['score'].min()),
    score_max=float(train_df['score'].max()),
).fit(train_df)

metrics = evaluate_cf(model, test_df, k=10, threshold=3.0)
print(metrics.round(6))
```
